```python
import pandas as pd
url = "https://raw.githubusercontent.com/D6200763-SUT/SUT_Rice_Project_2024/refs/heads/main/Import_Dataset/env_daily_with_rice_monthly_raw.csv"

df = pd.read_csv(url)
#df = pd.read_csv(url, parse_dates=["Date"])
#df = df.sort_values("Date")
print(df.head())
print(df.shape)

cols = df.columns.tolist()
print(cols)
```

```python
import numpy as np
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df = df.dropna(subset=["date"])

df["bph_raw"] = pd.to_numeric(df["bph_count"], errors="coerce").clip(lower=0)
df["bph_log1p"] = np.log1p(df["bph_raw"])

```

```python
df["month_num"] = df["date"].dt.month
df["doy"] = df["date"].dt.dayofyear

df["month_sin"] = np.sin(2*np.pi*df["month_num"]/12)
df["month_cos"] = np.cos(2*np.pi*df["month_num"]/12)
df["doy_sin"]   = np.sin(2*np.pi*df["doy"]/365)
df["doy_cos"]   = np.cos(2*np.pi*df["doy"]/365)
```

```python
ws = pd.to_numeric(df["wind_speed"], errors="coerce")
wd = pd.to_numeric(df["wind_direction"], errors="coerce")
theta = np.deg2rad(wd)

df["wind_u"] = ws * np.sin(theta)
df["wind_v"] = ws * np.cos(theta)
```

```python
df = df.sort_values(["station_id","date"])

df["delta_temp"] = df.groupby("station_id")["temp"].diff()
df["temp_7d_mean"] = df.groupby("station_id")["temp"].transform(
    lambda s: s.rolling(7, min_periods=1).mean()
)

df["humidity_7d_mean"] = df.groupby("station_id")["humidity"].transform(
    lambda s: s.rolling(7, min_periods=1).mean()
)

df["rain_7d_sum"] = df.groupby("station_id")["rainfall"].transform(
    lambda s: s.rolling(7, min_periods=1).sum()
)
```
